### Dataset Overview

**1) Market & Financial Data**

1A. Target oil price
- EIA Europe Brent Spot Price FOB

1B. Reference oil price
- EIA Cushing, OK WTI Spot Price FOB

1C. U.S. petroleum fundamentals
- EIA Weekly Petroleum Status Report
  - commercial crude stocks
  - Cushing crude stocks
  - crude production
  - crude imports / exports
  - refinery inputs
  - refinery utilisation
  - gasoline product supplied
  - distillate product supplied
  - jet fuel product supplied

1D. Macro-financial control variables
- FRED Nominal Broad U.S. Dollar Index
- FRED Effective Federal Funds Rate / DFF
- FRED 10-Year Treasury Yield / DGS10
- FRED / CBOE VIX
- Yahoo Finance S&P 500 (^GSPC)
